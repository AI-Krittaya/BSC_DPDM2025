#ระบุข้อมูลที่หายไป

```python
#ระบุข้อมูลสูญหาย (record ที่หายไป) ของแต่ละสถานี แสดงจำนวนและสัดส่วน
import pandas as pd

# 1. Prepare Data
daily_2025 = daily_2025.copy()
daily_2025['measure_datetime'] = pd.to_datetime(daily_2025['measure_datetime'])

# 2. Calculate Global Date Range constraints
min_date = daily_2025['measure_datetime'].min()
max_date = daily_2025['measure_datetime'].max()
# Calculate total days in the range (Total Possible Records)
total_possible_records = len(pd.date_range(start=min_date, end=max_date, freq='D'))

# 3. Core Logic: Use GroupBy instead of For Loop
# Group by station and count unique dates present for each
# (This replaces the merge inside the loop)
station_stats = daily_2025.groupby('tele_station_oldcode')['measure_datetime'].nunique().reset_index()
station_stats.columns = ['Station', 'Present_Records']

# 4. Calculate Missing Data
station_stats['Missing Records Count'] = total_possible_records - station_stats['Present_Records']
station_stats['Missing Records Proportion'] = station_stats['Missing Records Count'] / total_possible_records

# 5. Final Formatting to match original output exactly
missing_data_df = station_stats[['Station', 'Missing Records Count', 'Missing Records Proportion']]

# Display results
display(missing_data_df.head())
print(f"\nNumber of stations with missing data: {len(missing_data_df[missing_data_df['Missing Records Count'] > 0])}")
```

#คัดสถานีที่มีข้อมูลสูญหาย >=20% ออก

```python
#คัดสถานีที่มีข้อมูลสูญหาย >=20% ออก แสดงรายชื่อสถานีที่ถูกคัดออกและจำนวนที่เหลือ
# 1. Create a boolean mask for the condition (Reusable)
# สร้างเงื่อนไขแยกไว้ก่อน เพื่อนำไปใช้ทั้งการหาตัวลบ และหาตัวเก็บ
high_missing_mask = missing_data_df['Missing Records Proportion'] >= 0.20

# 2. Get the list of stations to remove using .loc
# ใช้ .loc เพื่อเข้าถึงข้อมูลที่ตรงตามเงื่อนไขทันที
removed_station_codes = missing_data_df.loc[high_missing_mask, 'Station'].tolist()

# 3. Print the list (Using join instead of loop)
# ใช้ .join() แทน for loop เพื่อรวบโค้ดให้สั้นลงและทำงานเร็วขึ้นสำหรับการแสดงผลข้อความ
print("รายชื่อสถานีที่ถูกคัดออกเนื่องจากมีข้อมูลสูญหาย >= 20%:")
if removed_station_codes:
    print("- " + "\n- ".join(removed_station_codes))
else:
    print("ไม่มีสถานีใดถูกคัดออกในเกณฑ์นี้")

print(f"\nจำนวนสถานีที่ถูกคัดออก: {len(removed_station_codes)}")

# 4. Filter daily_2025 using Positive Selection Logic
# นับจำนวนก่อนทำ
before_removal_count = daily_2025['tele_station_oldcode'].nunique()

# แทนที่จะหาตัวลบแล้วใช้เครื่องหมาย ~ (not) เราหา "รายชื่อสถานีที่ผ่านเกณฑ์" (Good Stations) แทน
# แล้วเลือกเฉพาะข้อมูลที่อยู่ในรายชื่อที่ดี
good_stations = missing_data_df.loc[~high_missing_mask, 'Station'] # สังเกตเครื่องหมาย ~ ที่กลับเงื่อนไขตรงนี้แทน
daily_2025 = daily_2025[daily_2025['tele_station_oldcode'].isin(good_stations)]

after_removal_count = daily_2025['tele_station_oldcode'].nunique()

print(f"จำนวนสถานีก่อนคัดออก: {before_removal_count}")
print(f"จำนวนสถานีที่เหลือหลังจากคัดออก: {after_removal_count}")
```

#เติมข้อมูลที่หายไปด้วยค่า Median


```python
#เติมข้อมูลที่หายไปด้วยค่า Median ตาม Year จาก historical แยกตามสถานี
# 1. Prepare Historical Median Data
historical['measure_datetime'] = pd.to_datetime(historical['measure_datetime'])

# --- แก้ไขจุดที่ทำให้เกิด Error: สร้างคอลัมน์ doy ให้ชัดเจนก่อน Group ---
historical['doy'] = historical['measure_datetime'].dt.dayofyear
median_ref = historical.groupby(['station_code', 'doy'])['data'].median().reset_index()
# ------------------------------------------------------------------

# Rename columns immediately (เตรียม Merge)
median_ref.rename(columns={'station_code': 'tele_station_oldcode', 'data': 'median_fill_value'}, inplace=True)

# -----------------------------------------------------------------------------
# Optimized Logic: Use 'Reindex' instead of 'Merge with Cartesian Product'
# -----------------------------------------------------------------------------

# 2. Setup the Full Grid (MultiIndex)
unique_stations = daily_2025['tele_station_oldcode'].unique()
date_range = pd.date_range(start=daily_2025['measure_datetime'].min(),
                           end=daily_2025['measure_datetime'].max(),
                           freq='D')

full_idx = pd.MultiIndex.from_product(
    [unique_stations, date_range],
    names=['tele_station_oldcode', 'measure_datetime']
)

# 3. Expand Data using Reindex
daily_2025_imputed = (
    daily_2025
    .drop(columns=['doy'], errors='ignore')
    .set_index(['tele_station_oldcode', 'measure_datetime'])
    .reindex(full_idx)
    .reset_index()
)

# 4. Impute Missing Values
# คำนวณ DOY ใหม่สำหรับข้อมูล daily_2025_imputed
daily_2025_imputed['doy'] = daily_2025_imputed['measure_datetime'].dt.dayofyear

# ตอนนี้ทั้ง daily_2025_imputed และ median_ref มีคอลัมน์ 'doy' ทั้งคู่แล้ว Merge ได้แน่นอน
daily_2025_imputed = pd.merge(
    daily_2025_imputed,
    median_ref,
    on=['tele_station_oldcode', 'doy'],
    how='left'
)

# Fillna: ใช้ค่าจาก median_fill_value ไปอุดรูรั่ว
daily_2025_imputed['rainfall_value'] = daily_2025_imputed['rainfall_value'].fillna(daily_2025_imputed['median_fill_value'])

# 5. Cleanup
daily_2025_imputed.drop(columns=['doy', 'median_fill_value'], inplace=True)

display(daily_2025_imputed.head())
```

## เหตุผลที่ใช้ Median แทน Mean
* **ข้อมูลปริมาณน้ำฝนมักมีการกระจายตัวที่ไม่สมมาตร** (Skewed Distribution): ข้อมูลปริมาณน้ำฝนส่วนใหญ่มักจะมีค่าเป็นศูนย์ (ไม่มีฝนตก) หรือมีค่าต่ำๆ จำนวนมาก และมีค่าสูงๆ (ฝนตกหนัก) เป็นส่วนน้อย ซึ่งทำให้การกระจายตัวของข้อมูลเอียงไปทางขวา (positively skewed).

* **ค่าเฉลี่ย** (Mean) อ่อนไหวต่อค่าผิดปกติ (Outliers) และการกระจายตัวที่ไม่สมมาตร: ถ้ามีค่าฝนตกหนักมากๆ เพียงไม่กี่วัน ค่าเฉลี่ยจะถูกดึงให้สูงขึ้นอย่างมาก ซึ่งอาจไม่สะท้อนถึงปริมาณน้ำฝน 'โดยทั่วไป' ที่สถานีนั้นๆ ได้รับ

* **ค่ามัธยฐาน** (Median) มีความทนทานต่อค่าผิดปกติ (Robust to Outliers): ค่ามัธยฐานคือค่ากลางที่แบ่งข้อมูลออกเป็นสองส่วนเท่าๆ กัน (50% ของข้อมูลมีค่าน้อยกว่าหรือเท่ากับค่ามัธยฐาน และ 50% มีค่ามากกว่าหรือเท่ากับ) การที่ข้อมูลมีค่าสูงหรือต่ำผิดปกติจะไม่ส่งผลกระทบต่อค่ามัธยฐานมากเท่ากับค่าเฉลี่ย ทำให้ค่ามัธยฐานเป็นตัวแทนของ 'ค่าทั่วไป' หรือ 'ค่าปกติ' ของปริมาณน้ำฝนได้ดีกว่าในสถานการณ์เช่นนี้

#คำนวณผลรวม

```python
#คำนวณผลรวมปริมาณน้ำฝนรายเดือน
# แปลงเป็น datetime ให้เรียบร้อย
daily_2025_imputed['measure_datetime'] = pd.to_datetime(daily_2025_imputed['measure_datetime'])

# เริ่ม Pipeline: สร้าง, รวม, เปลี่ยนชื่อ, จัดเรียง ในคำสั่งเดียว
output_monthly_rainfall = (
    daily_2025_imputed
    .assign(
        year=lambda x: x['measure_datetime'].dt.year,
        month=lambda x: x['measure_datetime'].dt.month
    )
    .groupby(['tele_station_oldcode', 'year', 'month'], as_index=False)
    .agg(rain=('rainfall_value', 'sum')) # Named Aggregation: รวมผลแล้วตั้งชื่อว่า 'rain' ทันที
    .rename(columns={'tele_station_oldcode': 'Node'})
    .sort_values(['Node', 'year', 'month'])
    [['Node', 'rain', 'month', 'year']] # จัดลำดับคอลัมน์ปิดท้าย
)

display(output_monthly_rainfall.head())
```
